### Create a string, sent = ‘Coronavirus Caused Lockdowns Around The World.”

```python

sent1 = sent.split()
sent1
```

Convert each word in the list to lower case and store it in variable words_lower.

```python
worlds_lower = [i.lower() for i in sent1]
worlds_lower
```

Check whether ‘country’ is in the list

```python
'country' in worlds_lower
```

Remove the element ‘the’ from the list and print the list.

```python
worlds_lower.remove('the')
worlds_lower
```

```python
# print x4
print(a)
print(b)
print(c)
print(d)
```

```python
sent_joined = ' '.join(worlds_lower)
sent_joined
```
